```python
df = df.drop(columns=['Unnamed: 0'])
```

```python
df.head()
```

```python
print("Average doctor visits:", round(df['visits'].mean(), 2))
print("Minimum doctor visits:", df['visits'].min())
print("Maximum doctor visits:", df['visits'].max())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))

sns.histplot(df['visits'], bins=20, kde=True)

plt.title("Distribution of Doctor Visits")
plt.xlabel("Number of Visits")
plt.ylabel("Number of People")

plt.show()
```

```python
print(df.groupby('gender')['visits'].mean())
```

```python
plt.figure(figsize=(7,5))

sns.barplot(data=df, x='gender', y='visits')

plt.title("Gender vs Doctor Visits")
plt.xlabel("Gender")
plt.ylabel("Average Doctor Visits")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='income',
    y='visits'
)

plt.title("Income vs Doctor Visits")
plt.xlabel("Income")
plt.ylabel("Doctor Visits")

plt.show()
```

```python
print(df.groupby('illness')['visits'].mean())
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x='illness',
    y='visits'
)

plt.title("Illness vs Doctor Visits")
plt.xlabel("Number of Illnesses")
plt.ylabel("Average Doctor Visits")

plt.show()
```

```python
print(df.groupby('health')['visits'].mean())
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x='health',
    y='visits'
)

plt.title("Health Status vs Doctor Visits")
plt.xlabel("Health Status")
plt.ylabel("Average Doctor Visits")

plt.show()
```

```python
print("NCHRONIC")
print(df.groupby('nchronic')['visits'].mean())

print("\nLCHRONIC")
print(df.groupby('lchronic')['visits'].mean())
```

```python
print("Private healthcare:")
print(df.groupby('private')['visits'].mean())

print("\nFree healthcare:")
print(df.groupby('freepoor')['visits'].mean())

print("\nRepatriation healthcare:")
print(df.groupby('freerepat')['visits'].mean())
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,7))

numeric_data = df.select_dtypes(include=np.number)

sns.heatmap(
    numeric_data.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Between Healthcare Variables")

plt.show()
```
